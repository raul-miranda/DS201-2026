## Part A -- Conceptual

For the R code below:

1. Predict the output
2. Translate it into Python
3. Compute the result in Python
  

x <- c(1, 3, 5, 7)
y <- ifelse(x > 4, x^2, x + 1)
mean(y)
  
### Answer:

1. The code compares each vector element with 4; if greater than 4, squares the element, otherwise it adds 1 to the element. Thus, the output will be the mean of the y vector elements: (2, 4, 25, 49): 20.
  

2. Translation into Python:
  

       import numpy as np
       x = np.array([1, 3, 5, 7])
       y = np.where(x > 4, x**2, x + 1)
       y_mean = np.mean(y)
       print(y_mean)
  

3. Compute in Python:

```python
import numpy as np
x = np.array([1, 3, 5, 7])
y = np.where(x > 4, x**2, x + 1)
y_mean = np.mean(y)
print(y_mean)
```

## Part B -- Lists vs Numpy Arrays

1. Create a Python list of numbers 1--5
2. Attempt to multiply the list by 2
3. Convert the list to a NumPy array and repeat the multiplication

### Answer:

```python
x = [1, 2, 3, 4, 5]
y = x * 2
print(y)

x_np = np.array(x) 
y_np = x_np * 2
print(y_np)
```

The Python list multiplied by 2 just repeats the list twice.
The Numpy array multiplied by 2 multiplies each element by 2.
The Numpy array acts like a vector, while the Python list is a container; lists don't support element-wise operations.

## Part C - Short Reflection

The key conceptual difference I saw was that R and Python work differently with lists and arithmetic operations. R lists operate like vectors and additions or multiplications are done elementwise. Python lists operate like containers: multiplication by an integer produces repetition of the list; addition of lists produce concatenation of them. However, Numpy arrays behave like vectors and it is possible to achieve similar behavior to R.
